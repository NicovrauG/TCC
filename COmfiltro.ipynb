```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

aaa = 'dados_esp32_2.csv'
df = pd.read_csv(aaa, dtype=str)

plt.figure(figsize=(30,15))
plt.plot(df['indice'], df['dado'], color='b', linewidth=0.5)
plt.title('Sinal do EMG')
plt.xlabel('Indice')
plt.ylabel('amplitude')
plt.grid(True)
plt.show()
```
